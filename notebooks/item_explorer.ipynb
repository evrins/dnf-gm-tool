```python
from collections import defaultdict
from typing import List

coll: dict[str, List[dict]] = defaultdict(list)
for it in (
    df.filter(pl.col('stackable_type') == '[enchant waste]').select('*').to_dicts()
):
    d = json.loads(it['json'])
    monster_card_id = d['[monster card id]']
    if not monster_card_id:
        print(f'monster card id not found in {it["id"]}')
    card = df.filter(pl.col('id') == monster_card_id[0]).to_dicts()[0]
    card_d = json.loads(card['json'])
    enchant_key = '[enchant]'
    if enchant_key not in card_d:
        print(f'enchant key {enchant_key} not found in card {card["name"]}')
        coll['others'].append(d)
        continue
    categories = card_d[enchant_key].keys()
    for c in categories:
        coll[c].append(d)

for k, v in coll.items():
    print(f'{k}')
    for d in v:
        print(f'\t{d}')
```
